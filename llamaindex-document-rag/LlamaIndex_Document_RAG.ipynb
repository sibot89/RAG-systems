# Document RAG with LlamaIndex

A simple document-based Retrieval-Augmented Generation (RAG) implementation using LlamaIndex, Hugging Face embeddings, and Google's Gemini model.

The workflow demonstrates document ingestion, vector indexing, retrieval, and answer generation through a high-level RAG framework.

## 2. Imports and Environment Configuration

```python
import os

from dotenv import load_dotenv
from llama_index.core import VectorStoreIndex, SimpleDirectoryReader, Settings
from llama_index.llms.gemini import Gemini
from llama_index.embeddings.huggingface import HuggingFaceEmbedding

load_dotenv()

api_key = os.getenv('GOOGLE_API_KEY')

if not api_key:
    raise ValueError('GOOGLE_API_KEY is not set. Add it to your .env file.')
```

Create a `.env` file containing:

`GOOGLE_API_KEY=your_api_key_here`

Do not upload the `.env` file to GitHub.

## 3. Configure the LLM and Embedding Model

```python
MODEL_NAME = 'gemini-2.5-flash'
EMBEDDING_MODEL = 'BAAI/bge-small-en-v1.5'

llm = Gemini(
    model=MODEL_NAME,
    api_key=api_key
)

embed_model = HuggingFaceEmbedding(
    model_name=EMBEDDING_MODEL
)

Settings.llm = llm
Settings.embed_model = embed_model
```

## 4. Load Documents

```python
document_dir = './data'

if not os.path.isdir(document_dir):
    raise FileNotFoundError(f'Document directory not found: {document_dir}')

documents = SimpleDirectoryReader(document_dir).load_data()

if not documents:
    raise ValueError('No documents were found in the data directory.')

print(f'Loaded {len(documents)} document(s).')
```

## 5. Build the Vector Index

```python
index = VectorStoreIndex.from_documents(documents)

print('Vector index created successfully.')
```

## 6. Query the RAG System

```python
query_engine = index.as_query_engine()

query = input('Enter your question: ').strip()

if not query:
    raise ValueError('Please enter a question.')

response = query_engine.query(query)

print('\n########## Answer ##########\n')
print(response)
```

## Pipeline

Documents → SimpleDirectoryReader → Hugging Face Embeddings → VectorStoreIndex → Retrieval → Gemini → Answer

## Notes and Limitations

This is an educational implementation intended to demonstrate the basic workflow of a framework-based RAG system.

- Uses LlamaIndex to manage document ingestion and indexing
- Uses `BAAI/bge-small-en-v1.5` for embeddings
- Uses a Gemini model for answer generation
- Requires a valid Google API key
- No persistent vector store configuration
- No reranking or metadata filtering
- Not intended as a production-ready RAG application
